```python
import nltk
import pandas as pd
import numpy as np
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python

# Load dataset
df = pd.read_csv("all-data.csv", encoding="ISO-8859-1")

# Convert sentiment labels to numerical values
df["Sentiment"] = df["Type"].map({"positive": 1, "neutral": 0, "negative": -1})

# Verify the mapping
print(df["Sentiment"].unique())  # Should print: [1, 0, -1]
```

#cleaning

```python
mydata=[]
for i in df["Message"]:
    text = re.sub(r',','',i)
    cleaned_text = re.sub(r'[^\w\s]', '', text)  # Removes everything except word characters and spaces
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text)  # Replaces multiple spaces with a single space
    cleaned_data = cleaned_text.strip()  # Removes leading/trailing whitespace
    mydata.append(cleaned_data)
```

```python
#to tokenize the whole data  :
#tokenization can't be done for the whole data at once
#tokenization should be done one message at a time

tokens = [] #create a list to store the tokens
for x in mydata:
    h=word_tokenize(x)
    tokens.append(h)

tokens
```

#removing stopwords

```python
#to see the stop words in english : stopwords.words('english')

stop = [] # list to store stop words

stop.append(stopwords.words('english'))
stop
```

```python
data=[]
for i in (tokens):
    if i not in stop:
        data.append(i)
data
```

#stemming

```python
# stemming : Stemming reduces words to their root form by removing suffixes or prefixes, often without considering the actual meaning of the word
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
#stemming should be done word by word : ps.stem(word)

stemdata =[] #creating an empty list to store stem data
for message in data:
    # for word in message:
    #     st=ps.stem(word)
    #     stemdata.append(st)
    st = [ps.stem(word) for word in message]
    stemdata.append(st)
stemdata
```

```python
from nltk.corpus import wordnet
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')  # For better language support
from nltk.stem import WordNetLemmatizer

# Initialize the lemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
# Function to convert nltk's POS tags to WordNet's POS tags:


def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ  # Adjective
    elif tag.startswith('V'):
        return wordnet.VERB  # Verb
    elif tag.startswith('N'):
        return wordnet.NOUN  # Noun
    elif tag.startswith('R'):
        return wordnet.ADV  # Adverb
    else:
        return wordnet.NOUN  # Default to noun if POS tag is unknown
```

```python
pos_token[0][5][1]
# [ message index] [row index in message] [ column index in row]
```

```python
lem_data = []
for x in range(len(pos_token)):
  lem = [lemmatizer.lemmatize(pos_token[x][i][0],get_wordnet_pos(pos_token[x][i][1])) for i in range(len(pos_token[x]))]
  lem_data.append(lem)
```

```python

# Features (X) and Labels (y)
X_text = df["Message"]  # News text (features)
y = df["Sentiment"]      # Sentiment labels (target)

# Train the vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(X_text)

# Train the model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X, y)

# Save vectorizer and model
import pickle

with open("vectorizer.pickle", "wb") as f:
    pickle.dump(vectorizer, f)

with open("model.pickle", "wb") as f:
    pickle.dump(model, f)

print("Vectorizer and model saved successfully!")
```
